# Interpretación Alpha de Cronbach

![Matriz de Correlación](tabla_cronbach.jpg)

## 1. Marco de Interpretación de las Métricas

+ **Alfa de Cronbach ($\alpha$):** Mide la consistencia interna (si los ítems de una dimensión apuntan hacia el mismo concepto teórico). Un estándar común en investigación social establece que $\alpha \ge 0.70$ es aceptable.

+ **Correlación Inter-Ítem Media:** Mide cuánto se relacionan las preguntas entre sí dentro de la misma dimensión. El rango ideal suele situarse entre $0.20$ y $0.50$ (o hasta $0.70$ en dimensiones de pocos ítems). Valores superiores a $0.80$ indican redundancia; valores cercanos a $0$ o negativos indican que las preguntas no miden lo mismo.

## 2. Análisis del Constructo A (Sí Rindieron, $N=9$)

**Dimensiones Sólidas y Confiables ($\alpha \ge 0.70$)**

+ **AD5_aceptacion ($\alpha = 0.852$, Corr = $0.769$):** Excelente consistencia. Los estudiantes responden de manera altamente coherente frente a la disposición de aceptar y seguir utilizando la retroalimentación asistida con IA.   

+ **AD1_retralim_util ($\alpha = 0.810$, Corr = $0.579$):** Alta consistencia y correlación media idónea. Los $3$ ítems asociados a la utilidad técnica, identificación de brechas y aporte al aprendizaje se comportan homogéneamente.   

+ **AD4_docen_comprometid ($\alpha = 0.728$, Corr = $0.439$):** Adecuada consistencia. Refleja que la recodificación de las preguntas invertidas funcionó correctamente para integrar el compromiso, control e innovación docente.   

**Dimensiones con Problemas Psicométricos ($\alpha < 0.70$)**

+ **AD2_evaluac_justa ($\alpha = 0.167$, Corr = $0.068$):** Consistencia casi nula. Las respuestas a los ítems sobre imparcialidad, objetividad y transparencia no covarían en el grupo. Esto suele ocurrir cuando los estudiantes perciben la transparencia de la pauta de forma distinta a la objetividad de la nota.   

+ **AD3_revisa_oportuna ($\alpha = -0.286$, Corr = $-0.125$):** Alfa negativo. Muestra una relación inversa entre los $2$ ítems de la dimensión (tiempo de entrega vs legibilidad del formato Word). Para los estudiantes, el beneficio del formato digital no está asociado conceptualmente a la rapidez de entrega.

## 3. Análisis del Constructo B (No Rindieron, $N=4$)

+ **BD3_confianza ($\alpha = 0.778$, Corr = $0.636$):** Aceptable. Existe consenso razonable entre el deseo de revisión asistida y la confianza en la supervisión docente.   

+ **BD1_expectativa ($\alpha = 0.750$, Corr = $1.0$):** Presenta un alfa aceptable, pero una correlación inter-ítem de $1.0$. En un tamaño muestral de $N=4$, esto señala que los respondedores marcaron patrones idénticos o proporcionales en ambos ítems de expectativa, reflejando respuesta determinista o redundancia.   

+ **BD2_docen_comprometid ($\alpha = 0.556$, Corr = $0.343$):** Consistencia insuficiente. Las expectativas previas sobre compromiso, control e innovación no se agrupan de forma compacta en la muestra que no rindió.   

![Matriz de Correlación](tabla_cronbach_no.jpg)

## 4. Diagnóstico

**1. Efecto del tamaño muestral ($N=9$ y $N=4$):** En muestras pequeñas, el Alfa de Cronbach es especialmente sensible a la varianza de cada ítem. La métrica confirma que el instrumento funciona para medir aceptación (AD5), utilidad percibida (AD1) y compromiso docente (AD4).   

**2. Ajuste metodológico necesario:** En el informe o artículo académico, conviene tratar AD3_tiempo y AD3_legibilidad como indicadores univariados independientes en lugar de promediarlos en un único constructo, dada su correlación negativa.

# Prueba de Hipótesis

**¿Es percibido el Docente como 'Comprometido con el éxito del curso'?**

Para responder formalmente a esta pregunta con un 95% de confianza ($\alpha = 0.05$), debemos realizar una prueba t de Student para una muestra (one-sample t-test) sobre la dimensión de compromiso docente.   Como la escala tipo Likert suele ser de 1 a 5 (donde el punto neutro es 3.0), la hipótesis contrasta si la media percibida ($\mu$) es significativamente superior a la neutralidad ($3.0$):

+ **$H_0$ (Hipótesis Nula): $\mu \le 3.0$** (El docente no es percibido como comprometido).

+ **$H_1$ (Hipótesis Alternativa): $\mu > 3.0$** (El docente es percibido significativamente como comprometido).
