# Lógica y Fundamento del Alfa de Cronbach

El Alfa de Cronbach ($\alpha$) es un coeficiente diseñado para medir la consistencia interna de un instrumento de medición o escala (por ejemplo, un cuestionario con escala Likert de 1 a 5).   

## 1. Qué mide en el contexto del script

+ Consistencia Interna / Unidimensionalidad: Evalúa en qué medida los ítems de una dimensión miden el mismo constructo teórico subyacente. Si las respuestas a las preguntas fluctúan de forma coordinada entre los encuestados, los ítems covarían y el índice es alto.

+ Reducción del Error de Medición: Toda respuesta observada contiene una medida verdadera y un error aleatorio. Un $\alpha$ elevado indica que la mayor parte de la varianza en los puntajes totales se debe a la señal común (el constructo) y no al ruido aleatorio.

+ Escala de Interpretación:

    + $\alpha \ge 0.9$: Excelente.
    
    + $0.8 \le \alpha < 0.9$: Bueno.
    
    + $0.7 \le \alpha < 0.8$: Aceptable (criterio utilizado en la función evaluar_validez_dimensiones).   
    
    + $\alpha < 0.7$: Cuestionable o bajo; sugiere reformular o eliminar ítems.   

## Fórmulas Matemáticas Explícitas

El cálculo del alfa se estructura en dos representaciones matemáticamente equivalentes:

Formulación Basada en Varianzas

$$\alpha = \left( \frac{K}{K - 1} \right) \left( 1 - \frac{\sum_{i=1}^{K} s_i^2}{s_T^2} \right)$$

Donde:$K$: Número de ítems de la dimensión.

$s_i^2$: Varianza muestral del ítem $i$.

$\sum_{i=1}^{K} s_i^2$: Suma de las varianzas individuales de cada ítem.

$s_T^2$: Varianza de las puntuaciones totales asignadas por sujeto (donde el total de cada sujeto es la suma de los puntajes obtenidos en los $K$ ítems).

## Formulación Estandarizada (Basada en Correlaciones)

Cuando las variables se estandarizan o presentan varianzas similares, el alfa se expresa en función del promedio de las correlaciones inter-ítem ($\bar{r}$), el cual el script calcula mediante el promedio del triángulo superior de la matriz de correlación (avg_corr):

   $$\alpha_{est} = \frac{K \cdot \bar{r}}{1 + (K - 1) \bar{r}}$$
   
## Ejemplo Paso a Paso

A partir de las dimensiones definidas en el script, se toma la dimensión AD5_aceptacion del Constructo A:   

+ Ítems ($K = 2$): AD5_comodidad ($X_1$) y AD5_seguir_usando ($X_2$).   

+ Muestra ($N = 4$ respuestas hipotéticas en escala Likert 1-5):

| Sujeto (j) | X1​ (comodidad) | X2​ (seguir_usando) | Puntaje Total (Tj​=X1​+X2​) |
| :--- | :--- | :--- | :--- |
| 1 | 2 | 2 | 4 |
| 2 | 3 | 4 | 7 |
| 3 | 4 | 4 | 8 | 
| 4 | 5 | 5 | 10 |

**Paso 1: Calcular Medias de cada Ítem y del Total**

$$\bar{X}_1 = \frac{2 + 3 + 4 + 5}{4} = 3.5$$

$$\bar{X}_2 = \frac{2 + 4 + 4 + 5}{4} = 3.75$$

$$\bar{T} = \frac{4 + 7 + 8 + 10}{4} = 7.25$$

**Paso 2: Calcular Varianzas Muestrales ($s^2 = \frac{\sum (X - \bar{X})^2}{N - 1}$)**

+ Varianza ítem 1 ($s_1^2$):

$$s_1^2 = \frac{(2-3.5)^2 + (3-3.5)^2 + (4-3.5)^2 + (5-3.5)^2}{4 - 1} = \frac{2.25 + 0.25 + 0.25 + 2.25}{3} = \frac{5}{3} \approx 1.667$$

+ Varianza ítem 2 ($s_2^2$):$$s_2^2 = \frac{(2-3.75)^2 + (4-3.75)^2 + (4-3.75)^2 + (5-3.75)^2}{3} = \frac{3.0625 + 0.0625 + 0.0625 + 1.5625}{3} = \frac{4.75}{3} \approx 1.583$$

+ Suma de varianzas individuales ($\sum s_i^2$):

$$\sum s_i^2 = 1.667 + 1.583 = 3.250$$

+ Varianza de las puntuaciones totales ($s_T^2$):

$$s_T^2 = \frac{(4-7.25)^2 + (7-7.25)^2 + (8-7.25)^2 + (10-7.25)^2}{3} = \frac{10.5625 + 0.0625 + 0.5625 + 7.5625}{3} = \frac{18.75}{3} = 6.250$$

**Paso 3: Aplicar la Fórmula del Alfa de Cronbach**

$$\alpha = \left( \frac{K}{K - 1} \right) \left( 1 - \frac{\sum s_i^2}{s_T^2} \right)$$

$$\alpha = \left( \frac{2}{2 - 1} \right) \left( 1 - \frac{3.250}{6.250} \right)$$

$$\alpha = 2 \cdot (1 - 0.520) = 2 \cdot 0.480 = 0.960$$

## Detalles del Cálculo en el Script

1. Inversión de Ítems Negativos:
Antes de calcular el alfa, el script invierte la escala de los ítems con redacción negativa (6 - valor) en la línea work_df.iloc[:, posiciones] = 6 - work_df.iloc[:, posiciones]. Si no se hiciese esto, los ítems negativos covariarían de manera inversa con los positivos, generando correlaciones negativas y distorsionando el alfa hacia valores artificialmente bajos o negativos.   

2. Filtrado Muestral:
Se aplican filtros por submuestras (df_A para $P0 = \text{'si'}$ y df_B para $P0 = \text{'no'}$) y se eliminan filas con datos nulos (dropna()).   

3. Llamada Automática:
La función pg.cronbach_alpha(data=df_sub) ejecuta internamente la relación entre la suma de las varianzas y la varianza del total sobre la matriz limpia de datos.
